# 10. Data Visualization

Employees By Department

```python
df.Department.value_counts().plot(kind="bar")
plt.title("Employees by Department")
plt.show()
```

Attrition Distribution

```python
df.Attrition.value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.title("Attrition Distribution")
plt.show()
```

Salary DistributionS

```python
df.MonthlyIncome.plot(kind ="hist")
plt.title("Monthly Income Distribution")
plt.show()
```

Age vs Monthly Income

```python
df.plot(x="Age", y="MonthlyIncome", kind="scatter")
plt.title("Age vs Monthly Income")
plt.show()
```

Top 5 Monthly Income By Department

```python
df.groupby("Department")["MonthlyIncome"].sum().sort_values(ascending=True).tail(5).plot(kind="barh")
plt.title("Monthly Income by Department")
plt.show()
```
